```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("Estudiantes_SVD.csv")
data
```

```python
nombres = data["Nombre"]

A = data.drop(["Nombre"], axis=1)
A
```

```python
from numpy import diag
from numpy import zeros
from scipy.linalg import svd

U, s, VT = svd(A)

Sigma = zeros((A.shape[0], A.shape[1]))
Sigma[:A.shape[1], :A.shape[1]] = diag(s)
```

```python
print("Valores singulares:")
print(s)
```

```python
n_elements = 2

UReduced2 = U[:, :n_elements]

SigmaReduced2 = Sigma[:n_elements, :n_elements]

VTReduced2 = VT[:n_elements, :]
```

```python
print("U reducida:")
print(UReduced2)

print("\nSigma reducida:")
print(SigmaReduced2)

print("\nVT reducida:")
print(VTReduced2)
```

```python
AReduced2 = UReduced2.dot(SigmaReduced2.dot(VTReduced2))

print("Matriz A transformada:")
print(AReduced2)
```

```python
T2 = UReduced2.dot(SigmaReduced2)

print("Matriz T (2 Dimensiones):")
print(T2)
```

```python
plt.figure(figsize=(8,6))

plt.scatter(T2[:,0], T2[:,1])

for i, nombre in enumerate(nombres):
    plt.annotate(nombre, (T2[i,0], T2[i,1]))

plt.title("SVD en 2 dimensiones")
plt.xlabel("Dimensión 1")
plt.ylabel("Dimensión 2")
plt.grid()

plt.show()
```

```python
n_elements = 3

UReduced3 = U[:, :n_elements]

SigmaReduced3 = Sigma[:n_elements, :n_elements]

VTReduced3 = VT[:n_elements, :]

AReduced3 = UReduced3.dot(SigmaReduced3.dot(VTReduced3))

print("Matriz A transformada:")
print(AReduced3)
```

```python
T3 = UReduced3.dot(SigmaReduced3)

print("Matriz T (3 Dimensiones):")
print(T3)
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(9,7))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(T3[:,0], T3[:,1], T3[:,2])

for i, nombre in enumerate(nombres):
    ax.text(T3[i,0], T3[i,1], T3[i,2], nombre)

ax.set_title("SVD en 3 dimensiones")
ax.set_xlabel("Dimensión 1")
ax.set_ylabel("Dimensión 2")
ax.set_zlabel("Dimensión 3")

plt.show()
```

```python
# Error con 2 dimensiones
Error2 = A.to_numpy() - AReduced2

plt.figure(figsize=(10,7))

dfError2 = pd.DataFrame(
    Error2,
    columns=["Matematicas", "Ciencias", "Espanol", "Historia", "EdFisica"]
)

dfError2.boxplot()
plt.title("Error de reconstrucción - 2 dimensiones")
plt.show()
```

```python
print("Errores promedio por materia:")
print(Error2.mean(axis=0))
```

```python
# Error con 3 dimensiones
Error3 = A.to_numpy() - AReduced3

plt.figure(figsize=(10,7))

dfError3 = pd.DataFrame(
    Error3,
    columns=["Matematicas", "Ciencias", "Espanol", "Historia", "EdFisica"]
)

dfError3.boxplot()
plt.title("Error de reconstrucción - 3 dimensiones")
plt.show()
```

```python
print("Errores promedio por materia:")
print(Error3.mean(axis=0))
```
